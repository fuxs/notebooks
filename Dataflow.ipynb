```python
import apache_beam as beam
import re
from apache_beam.io.textio import ReadFromText
from apache_beam.options.pipeline_options import PipelineOptions
import apache_beam.runners.interactive.interactive_beam as ib
from apache_beam.transforms.combiners import Count
```

Let's create our first simple pipeline. We use the `InteractiveRunner` to get more insights. Please ignore the `flags=[]` option, it helps to avoid wrong inputs from notebook.

```python
options = PipelineOptions(runner='InteractiveRunner',flags=[])
p = beam.Pipeline(options=options)
```

The pipeline consists of 2 transformations:
* [`Create`](https://beam.apache.org/releases/pydoc/current/apache_beam.transforms.core.html#apache_beam.transforms.core.Create) accepts an iterable, in this case a list with 3 strings, and returns a `PCollection`
* [`Map`](https://beam.apache.org/releases/pydoc/current/apache_beam.transforms.core.html#apache_beam.transforms.core.Map) accepts a function and calls it for each value of the input `PCollection`

```python
result = p | beam.Create(["dog", "cat", "mouse"]) | beam.Map(print)
```

```python
ib.show(result)
```

```python
ib.show_graph(p)
```

```python
def split_words(text):
    # convert the text string to a list of words
    words = re.findall(r"[\w\']+", text)
    # return a list of key/value tuples
    return [(x, 1) for x in words]


p2 = beam.Pipeline(options=options)
result2 = (
    p2
    | ReadFromText("gs://dataflow-samples/shakespeare/othello.txt")
    | beam.FlatMap(split_words)
    | Count.PerKey()
)
ib.show(result2)
```
